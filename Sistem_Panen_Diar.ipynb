```python
# SISTEM PENCATATAN HASIL PANEN DIGITAL

print("=== SISTEM PENCATATAN HASIL PANEN DIGITAL ===")

# Memasukkan data
nama_petani = input("Nama petani: ")
komoditas = input("Jenis komoditas: ")
jumlah_panen = float(input("Jumlah hasil panen (kg): "))
harga_per_kg = float(input("Harga jual per kg (Rp): "))

# Menghitung total nilai hasil panen
total_nilai = jumlah_panen * harga_per_kg

# Menampilkan hasil
print("\n=== DATA HASIL PANEN ===")
print("Nama petani       :", nama_petani)
print("Komoditas         :", komoditas)
print("Jumlah panen      :", jumlah_panen, "kg")
print("Harga per kg      : Rp", harga_per_kg)
print("Total nilai panen : Rp", total_nilai)

print("\nData hasil panen berhasil dicatat!")
```
